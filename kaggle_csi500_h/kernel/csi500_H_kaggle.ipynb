# CSI500 H seeds 1-3 (new_LiMT formal config)

Dataset `csi500-newlimt` attached, GPU on. The dataset root is located dynamically
(the Kaggle image nests mounts under /kaggle/input/datasets/<owner>/<slug>).
Resumable per seed; the results zip is refreshed after every seed.

```python
import glob, os, subprocess, sys
root = os.path.dirname(glob.glob('/kaggle/input/**/run_h_csi500_seed.py', recursive=True)[0])
print('dataset root:', root, flush=True)
work = '/kaggle/working'
os.makedirs(work + '/w0w2/data/v2', exist_ok=True)
os.makedirs(work + '/runs_kaggle', exist_ok=True)
for src_name, dst_name in [('new_v2_panel_csi500.npz', 'new_v2_panel_csi300.npz'),
                           ('new_v2_labels_csi500.npz', 'new_v2_labels_csi300.npz'),
                           ('new_v2_splits_csi500.npz', 'new_v2_splits_csi300.npz')]:
    dst = work + '/w0w2/data/v2/' + dst_name
    if not os.path.exists(dst):
        os.symlink(os.path.join(root, src_name), dst)
print('symlinks ready', flush=True)
result = subprocess.run([sys.executable, os.path.join(root, 'run_h_csi500_seed.py'),
                         '--root', work + '/w0w2', '--out-root', work + '/runs_kaggle',
                         '--seeds', '1', '2', '3'])
print('runner exit code:', result.returncode, flush=True)
if result.returncode != 0:
    raise SystemExit(result.returncode)
```
